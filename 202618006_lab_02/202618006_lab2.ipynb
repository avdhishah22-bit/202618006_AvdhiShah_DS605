## Task 2 - Vectorized Arithmetic and Linear Algebra

This task demonstrates vectorized arithmetic and basic linear algebra operations using NumPy matrices.

Matrix addition, element-wise multiplication, matrix multiplication, transpose, determinant, inverse, and verification of the inverse.

```python

# VECTORIZE ARITHMETIC AND LINEAR ALGEBRA



# ------------------------------------------------------------
# 1. Create two matrices
# ------------------------------------------------------------

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)


```

```python
# ------------------------------------------------------------
# 2. Matrix Addition
# ------------------------------------------------------------

addition = A + B

print("\nMatrix Addition (A + B):")
print(addition)


```

```python
# ------------------------------------------------------------
# 3. Element-wise Multiplication
# ------------------------------------------------------------

element_multiplication = A * B

print("\nElement-wise Multiplication (A * B):")
print(element_multiplication)


```

```python
# ------------------------------------------------------------
# 4. Matrix Multiplication
# ------------------------------------------------------------

matrix_multiplication = A @ B

print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)


# You can also use np.matmul()
matrix_multiplication2 = np.matmul(A, B)

print("\nMatrix Multiplication using np.matmul():")
print(matrix_multiplication2)


```

```python
# ------------------------------------------------------------
# 5. Transpose of a Square Matrix
# ------------------------------------------------------------

transpose_A = A.T

print("\nTranspose of Matrix A:")
print(transpose_A)


```

```python
# ------------------------------------------------------------
# 6. Determinant of Matrix A
# ------------------------------------------------------------

determinant_A = np.linalg.det(A)

print("\nDeterminant of Matrix A:")
print(determinant_A)


```

```python
# ------------------------------------------------------------
# 7. Check if Matrix A is Invertible
# ------------------------------------------------------------

if not np.isclose(determinant_A, 0):

    # Find inverse
    inverse_A = np.linalg.inv(A)

    print("\nInverse of Matrix A:")
    print(inverse_A)

    # --------------------------------------------------------
    # 8. Verify A @ A^-1 = Identity Matrix
    # --------------------------------------------------------

    identity_result = A @ inverse_A

    print("\nA @ A^-1:")
    print(identity_result)

    # Expected identity matrix
    identity_matrix = np.eye(A.shape[0])

    print("\nIdentity Matrix:")
    print(identity_matrix)

    # Verify using np.allclose()
    verification = np.allclose(identity_result, identity_matrix)

    print("\nIs A @ A^-1 equal to Identity Matrix?")
    print(verification)

else:
    print("\nMatrix A is not invertible because its determinant is zero.")
```

## Task 3 - Normal Distribution and Histogram

This task demonstrates how NumPy can be used to generate random values following a normal distribution.

At least 1,000 values are generated using a chosen mean and standard deviation. The sample statistics are calculated and compared with the chosen parameters, and a histogram is used to visualize the distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# NORMAL DISTRIBUTION AND HISTOGRAM
# ============================================================

# ------------------------------------------------------------
# 1. Generate values from a normal distribution
# ------------------------------------------------------------

# Set seed for reproducibility
np.random.seed(42)

# Chosen parameters
chosen_mean = 50
chosen_std = 10
n = 1000

# Generate 1000 values from a normal distribution
data = np.random.normal(
    loc=chosen_mean,
    scale=chosen_std,
    size=n
)

print("Number of values generated:", len(data))
print("Chosen mean:", chosen_mean)
print("Chosen standard deviation:", chosen_std)


```

```python
# ------------------------------------------------------------
# 2. Calculate sample mean and sample standard deviation
# ------------------------------------------------------------

sample_mean = np.mean(data)

# ddof=1 calculates SAMPLE standard deviation
sample_std = np.std(data, ddof=1)

print("\nSample Statistics:")
print("Sample mean:", sample_mean)
print("Sample standard deviation:", sample_std)


```

```python
# ------------------------------------------------------------
# 3. Compare sample values with chosen values
# ------------------------------------------------------------

mean_difference = sample_mean - chosen_mean
std_difference = sample_std - chosen_std

print("\nComparison:")
print("Difference in mean:", mean_difference)
print("Difference in standard deviation:", std_difference)


```

```python
# ------------------------------------------------------------
# 4. Plot histogram
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=30,
    edgecolor="black"
)

plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

## Task 5 - Filtering and Querying

Boolean indexing and Pandas filtering techniques are used to answer specific questions about the Titanic passengers.

The filtering conditions combine comparison operators and logical operators such as `&` and `|`.

```python
# FILTERING AND QUERYING

# ------------------------------------------------------------
# Male passengers older than 50
# ------------------------------------------------------------

male_above_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

print("1. Male passengers older than 50:")
print("Count:", len(male_above_50))
```

```python
# ------------------------------------------------------------
#  Female first-class passengers
#  How many and what percentage survived?
# ------------------------------------------------------------

female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

female_first_class_count = len(female_first_class)

female_first_class_survival_percentage = (
    female_first_class["Survived"].mean() * 100
)

print("\n2. Female first-class passengers:")
print("Count:", female_first_class_count)
print(
    "Percentage survived:",
    female_first_class_survival_percentage,
    "%"
)

```

```python
# ------------------------------------------------------------
#  Age 20-40, Fare above overall median, and survived
# ------------------------------------------------------------

overall_median_fare = df["Fare"].median()

age_20_40_high_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, fare above overall median, and survived:")
print("Overall median fare:", overall_median_fare)
print("Count:", len(age_20_40_high_fare_survived))
```

```python
# ------------------------------------------------------------
# Travelling alone, age < 30, and did not survive
# ------------------------------------------------------------

travelling_alone = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:")
print("Count:", len(travelling_alone))


```

```python
# ------------------------------------------------------------
#  Embarked = S, Pclass 2 or 3,
# Fare above Southampton median
# ------------------------------------------------------------

southampton_median_fare = df[
    df["Embarked"] == "S"
]["Fare"].median()

southampton_filtered = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print("\n5. Embarked = S, Pclass 2 or 3,")
print("   Fare above Southampton median:")
print("Southampton median fare:", southampton_median_fare)
print("Count:", len(southampton_filtered))
```

## Task 6 - groupby() and Aggregation

Pandas `groupby()` and aggregation functions are used to analyze survival patterns across different passenger groups.

The survival rate, passenger count, average age, and average fare are calculated for different categories.

```python
import pandas as pd

#GROUPBY() AND AGGREGATION

# ------------------------------------------------------------
# Survival rate by Sex
# ------------------------------------------------------------

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("1. Survival Rate by Sex:")
print(survival_by_sex)


```

```python
# ------------------------------------------------------------
# Survival rate by Pclass
# ------------------------------------------------------------

survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival Rate by Pclass:")
print(survival_by_pclass)


```

```python
# ------------------------------------------------------------
# Average Age and Fare by Pclass
# ------------------------------------------------------------

average_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(average_age_fare)


```

```python
# ------------------------------------------------------------
# Passenger count and survival rate by Sex-Pclass
# ------------------------------------------------------------

sex_pclass = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

# Convert survival rate to percentage
sex_pclass["Survival_Rate"] = sex_pclass["Survival_Rate"] * 100

print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_pclass)


```

```python
# ------------------------------------------------------------
# Passenger count, average Fare, and survival rate by Embarked
# ------------------------------------------------------------

embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

# Convert survival rate to percentage
embarked_stats["Survival_Rate"] = embarked_stats["Survival_Rate"] * 100

print("\n5. Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_stats)
```

## Task 7 - Missing Values and Fare Outliers

This task focuses on identifying and handling missing values and detecting potential outliers in the Fare column.

Different imputation techniques are compared, and the IQR method is used to identify Fare outliers.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#  MISSING VALUES AND FARE OUTLIERS
# ============================================================
#  Missing-value count and percentage for every column
# ============================================================

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing Values Summary:")
print(missing_summary)


```

```python
# ------------------------------------------------------------
# Plot missing-value counts
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


```

```python
# ============================================================
# Fill missing Age values with mean Age
# ============================================================

# Show missing Age values BEFORE filling
print("\nMissing Age values BEFORE filling:")
print(df["Age"].isnull().sum())

# Calculate mean Age
mean_age = df["Age"].mean()

print("Mean Age:", mean_age)

# Fill missing Age values with mean
df["Age"] = df["Age"].fillna(mean_age)

# Show missing Age values AFTER filling
print("\nMissing Age values AFTER filling:")
print(df["Age"].isnull().sum())


```

```python
# ============================================================
# Impute NaN values using mean, median, mode, and random
# ============================================================

# Create a separate copy so the original dataframe is not
# changed by the different imputation methods.

imputation_df = pd.read_csv("train.csv")



# Mean Imputation


mean_imputed = imputation_df["Age"].fillna(
    imputation_df["Age"].mean()
)

print("\nMean Imputation:")
print("Missing values:", mean_imputed.isnull().sum())



# Median Imputation


median_imputed = imputation_df["Age"].fillna(
    imputation_df["Age"].median()
)

print("\nMedian Imputation:")
print("Missing values:", median_imputed.isnull().sum())



# Mode Imputation


mode_age = imputation_df["Age"].mode()[0]

mode_imputed = imputation_df["Age"].fillna(mode_age)

print("\nMode Imputation:")
print("Mode Age:", mode_age)
print("Missing values:", mode_imputed.isnull().sum())



# Random Value Imputation


random_imputed = imputation_df["Age"].copy()

missing_indices = random_imputed[random_imputed.isnull()].index

# Set seed so random values are reproducible
np.random.seed(42)

# Select random existing Age values
random_values = np.random.choice(
    imputation_df["Age"].dropna(),
    size=len(missing_indices)
)

# Replace missing values with randomly selected Age values
random_imputed.loc[missing_indices] = random_values

print("\nRandom Value Imputation:")
print("Missing values:", random_imputed.isnull().sum())


```

```python
# ============================================================
# Fare Outliers using IQR Method
# ============================================================

# Calculate Q1 and Q3
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# Identify outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

# Count outliers
outlier_count = len(fare_outliers)



# Display Fare outlier results


print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", outlier_count)


```

## Task 8 - Features and Pivot Table

New features are created from existing Titanic dataset columns to provide additional information about passenger family structure.

A pivot table is then created to compare survival rates across Sex and Pclass.

```python
import pandas as pd

# FEATURES AND PIVOT TABLE

df = pd.read_csv("train.csv")


# ============================================================
# Create FamilySize
# ============================================================

# FamilySize = SibSp + Parch + 1
# +1 represents the passenger themselves

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print("FamilySize:")
print(df[["SibSp", "Parch", "FamilySize"]].head())


```

```python
# ============================================================
# 2. Create IsAlone
# ============================================================

# IsAlone = 1 if FamilySize is 1
# IsAlone = 0 otherwise

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("\nIsAlone:")
print(df[["FamilySize", "IsAlone"]].head())


```

```python
# ============================================================
# 3. Create Pivot Table
# ============================================================

pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("\nPivot Table - Mean Survival Rate:")
print(pivot_table)


```

```python
# ============================================================
# 4. Find Highest and Lowest Survival Groups
# ============================================================

# Convert pivot table into a Series so that we can
# easily find the highest and lowest values

survival_groups = pivot_table.stack()

highest_group = survival_groups.idxmax()
highest_rate = survival_groups.max()

lowest_group = survival_groups.idxmin()
lowest_rate = survival_groups.min()


print("\nHighest Survival Group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate * 100, "%")


print("\nLowest Survival Group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate * 100, "%")
```

## Task 9 - Visualizations and Observations

This task focuses on visualizing important relationships in the Titanic dataset and interpreting the results.

A correlation heatmap, survival rate bar chart, and Age vs Fare scatter plot are created. Numerical results are also used to support the final observations.

```python

# VISUALIZATIONS AND OBSERVATIONS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# ============================================================
#  Correlation Heatmap
# ============================================================

# Select relevant numerical columns
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation_matrix = df[numerical_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Numerical Features")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Find strongest positive and negative relationships
# ------------------------------------------------------------

# Remove self-correlations (diagonal values = 1)
corr_without_diagonal = correlation_matrix.copy()

np.fill_diagonal(corr_without_diagonal.values, np.nan)

# Find strongest positive correlation
positive_pair = corr_without_diagonal.stack().idxmax()
positive_value = corr_without_diagonal.stack().max()

# Find strongest negative correlation
negative_pair = corr_without_diagonal.stack().idxmin()
negative_value = corr_without_diagonal.stack().min()

print("\nStrongest Positive Relationship:")
print(positive_pair, "=", positive_value)

print("\nStrongest Negative Relationship:")
print(negative_pair, "=", negative_value)


```

```python
# ============================================================
# 2. Survival Rate by Sex
# ============================================================

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival Rate by Sex:")
print(survival_by_sex)

plt.figure(figsize=(7, 5))

survival_by_sex.plot(kind="bar")

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


```

```python
# ============================================================
# 3. Age vs Fare - Distinguished by Survival
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")

plt.tight_layout()
plt.show()


```

```python
# ============================================================
# 4. Additional Numerical Results for Observations
# ============================================================

# Overall survival rate
overall_survival_rate = df["Survived"].mean() * 100

# Survival rate by class
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

# Average fare of survivors and non-survivors
average_fare_survival = df.groupby("Survived")["Fare"].mean()

# Average age of survivors and non-survivors
average_age_survival = df.groupby("Survived")["Age"].mean()

# Passenger count by sex
passenger_count_sex = df["Sex"].value_counts()

print("\nAdditional Results:")
print("Overall Survival Rate:", overall_survival_rate, "%")

print("\nSurvival Rate by Pclass:")
print(survival_by_class)

print("\nAverage Fare by Survival:")
print(average_fare_survival)

print("\nAverage Age by Survival:")
print(average_age_survival)

print("\nPassenger Count by Sex:")
print(passenger_count_sex)
```
